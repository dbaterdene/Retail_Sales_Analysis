```python
#   ========================================================
#   Import Libraries    
#   ========================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns # Added for enhanced visual aesthetics
import warnings

#   ========================================================
#   # Suppress warnings for a cleaner notebook presentation   
#   ========================================================

warnings.filterwarnings('ignore')
```

```python
#   ========================================================
#   Load the Superstore sales dataset   
#   ========================================================

file_path = "/Users/golden_horde/Desktop/DA_projects/Retail-Sales-Performance/Data/Superstore.csv" 
df = pd.read_csv(file_path, encoding_errors='ignore') 

#   ========================================================
#   Display the first few rows to verify successful loading
#   ========================================================

df.head()
```

```python
#   ========================================================
# Inspect the dataset structure, correct data types, and non-null counts
#   ========================================================

df.info() 
```

```python
#   ========================================================
# Generate descriptive statistics for numerical columns
#   ========================================================

df.describe()
```

```python
#   ========================================================
# Check for missing values
#   ========================================================

missing_values = df.isnull().sum()
```

```python
#   ========================================================
# Convert date columns to proper datetime objects
#   ========================================================

df['Order Date'] = pd.to_datetime(df['Order Date']) 
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
#   ========================================================
# Identify and remove duplicate records
#   ========================================================

duplicate_count = df.duplicated().sum() 
print(f"Duplicate rows found: {duplicate_count}")

df = df.drop_duplicates()
```

```python
#   ========================================================
# Calculate core business KPIs
#   ========================================================

total_sales = df['Sales'].sum() 
total_profit = df['Profit'].sum() 
total_orders = df['Order ID'].nunique() 

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Unique Orders: {total_orders}")
```

```python
#   ========================================================
# Group sales by month-year to observe trends over time
#   ========================================================

df['Month-Year'] = df['Order Date'].dt.to_period('M')
monthly_sales = df.groupby('Month-Year')['Sales'].sum()

plt.figure(figsize=(14, 6))
monthly_sales.plot(kind='line', marker='o', color='b')
plt.title('Monthly Sales Trend', fontsize=14, fontweight='bold')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Total Sales ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
#   ========================================================
# Identify the Top 10 products generating the most revenue
#   ========================================================

top_products = (
    df.groupby('Product Name')['Sales'] 
      .sum() 
      .sort_values(ascending=False) 
      .head(10) 
)

```

```python
#   ========================================================
# Visualize Top 10 Products
#   ========================================================

plt.figure(figsize=(10, 6))
top_products.sort_values().plot(kind='barh', color='teal')
plt.title('Top 10 Products by Total Sales', fontsize=14, fontweight='bold')
plt.xlabel('Sales ($)')
plt.ylabel('Product Name')
plt.show()
```

```python
#   ========================================================
# Analyze sales distribution by Region
#   ========================================================

sales_region = (
    df.groupby('Region')['Sales'] 
      .sum() 
      .sort_values() 
)

plt.figure(figsize=(8, 5))
sales_region.plot(kind='barh', color='coral')
plt.title('Total Sales by Region', fontsize=14, fontweight='bold') 
plt.xlabel('Sales ($)')
plt.ylabel('Region')
plt.savefig('../images/sales_by_region.png', bbox_inches='tight') 
plt.show()
```
